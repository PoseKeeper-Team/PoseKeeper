```python
# 학습 결과를 Drive 에 저장
import shutil, os
DRIVE_DIR = '/content/drive/MyDrive/PoseKeeper/weights'
os.makedirs(DRIVE_DIR, exist_ok=True)
for name in ('mlp.pth', 'lstm.pth', 'autoencoder.pth'):
    src = f'weights/{name}'
    if os.path.exists(src):
        shutil.copy2(src, DRIVE_DIR)
        print('saved', src, '->', DRIVE_DIR)
```
